## Importing the Package

```python
from lawsuit_parser import __version__
from lawsuit_parser.utils import fetch_from_postgres, make_engine

print(f"lawsuit_parser version: {__version__}")
```

## Database Connection Example

Test the database connection and run a simple query.

```python
# Test database connection
engine = make_engine()
try:
    with engine.connect() as conn:
        result = conn.execute("SELECT version()")
        print(f"Connected to database: {result.scalar()}")
finally:
    engine.dispose()
```

## Fetching Data with Caching

The `fetch_from_postgres` function automatically caches query results.

```python
# Example query - adjust to match your actual database schema
query = """
SELECT 
    schemaname, 
    tablename 
FROM pg_tables 
WHERE schemaname NOT IN ('pg_catalog', 'information_schema')
LIMIT 10
"""

df = fetch_from_postgres(query)
print(f"Found {len(df)} tables")
df
```

## Working with DataFrames

Query results are returned as pandas DataFrames.

```python
import pandas as pd

# Example: Display table info
print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
df.head()
```

## Notes

- Query results are automatically cached in `data/cache/` based on query hash
- Use `force_refresh=True` to bypass the cache
- Use `output_filename` to save results to `data/` directory
